```python
import  matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('titles.csv')
df
```

```python
df['type'].value_counts().reset_index()
```

```python
df['type'].value_counts().plot(kind='bar')
plt.title('movies vs shows')
plt.xlabel('type')
plt.ylabel('count')
plt.show()
```

```python
movies=df[df['type']=='MOVIE']
```

```python
yearly_movies=movies.groupby('release_year')['title'].count()
yearly_movies.plot(kind='line',figsize=(11,5))
plt.title('movies released by year')
plt.xlabel('release year')
plt.ylabel('number of movies')
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.hist(df['imdb_score'].dropna(), bins=20)
plt.title('imdb score distribution')
plt.xlabel('imdb score')
plt.ylabel('frequency')
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.scatter(df['imdb_votes'],df['imdb_score'])
plt.title('imdb votes vs imdb score')
plt.xlabel('imdb votes')
plt.ylabel('imdb score')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df,x='type')
plt.title('number of movies and shows')
plt.xlabel('type')
plt.ylabel('count')
plt.show()
```
